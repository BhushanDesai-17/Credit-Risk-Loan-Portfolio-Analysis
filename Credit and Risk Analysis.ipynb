```python
import pandas as pd
```

```python
file = "credit_risk_loan_portfolio_dirty_data.xlsx"

loan_application = pd.read_excel(file, sheet_name = "Loan_Applications")
loan_application
```

```python
# Identifying NULL values

loan_application.isnull().sum()
```

```python
# Handling NULL Values

loan_application['Branch'] = loan_application['Branch'].fillna(loan_application['Branch'].mode()[0])

loan_application['Loan_Type'] = loan_application['Loan_Type'].fillna(loan_application['Loan_Type'].mode()[0])

loan_application['Employment_Type'] = loan_application['Employment_Type'].fillna(loan_application['Employment_Type'].mode()[0])

loan_application['Annual_Income_INR'] = loan_application['Annual_Income_INR'].fillna(loan_application['Annual_Income_INR'].median())

loan_application['Credit_Score'] = pd.to_numeric(loan_application['Credit_Score'], errors = 'coerce')
loan_application['Credit_Score'] = loan_application['Credit_Score'].fillna(loan_application['Credit_Score'].median())

loan_application['Debt_to_Income_Ratio'] = pd.to_numeric(loan_application['Debt_to_Income_Ratio'], errors = 'coerce')
loan_application['Debt_to_Income_Ratio'] = loan_application['Debt_to_Income_Ratio'].fillna(loan_application['Debt_to_Income_Ratio'].median())

loan_application['Interest_Rate_Pct'] = pd.to_numeric(loan_application['Interest_Rate_Pct'], errors = 'coerce')
loan_application['Interest_Rate_Pct'] = loan_application['Interest_Rate_Pct'].fillna(loan_application['Interest_Rate_Pct'].median())

loan_application['Tenure_Months'] = loan_application['Tenure_Months'].fillna(loan_application['Tenure_Months'].median())

loan_application['Application_Status'] = loan_application['Application_Status'].fillna(loan_application['Application_Status'].mode()[0])

loan_application
```

```python
# Droping Duplicates

loan_application.drop_duplicates(inplace = True)
loan_application
```

```python
loan_application.duplicated().sum()
```

```python
# Handling Inconsistent Data

loan_application['Branch'] = loan_application['Branch'].replace({
    'Aurangabad': 'Aurangabad',
    'Mumbai': 'Mumbai',
    'NAGPUR': 'Nagpur',
    'Nashik': 'Nashik',
    'Pune': 'Pune'
})

loan_application['Loan_Type'] = loan_application['Loan_Type'].replace({
    'Auto': 'Auto',
    'Auto Loan': 'Auto Loan',
    'Business': 'Business',
    'Education': 'Education',
    'HOME': 'Home',
    'personal': 'Personal'
})

loan_application['Employment_Type'] = loan_application['Employment_Type'].replace({
    'Business Owner': 'Business Owner',
    'salaried': 'Salaried',
    'Self-employed': 'Self-Employed',
    'Unemployed': 'Unemployed'
})

loan_application['Application_Status'] = loan_application['Application_Status'].replace({
    'Approved': 'Approved',
    'In Process': 'In Process',
    'Pending': 'Pending',
    'REJECTED': 'Rejected'
})

loan_application
```

```python
loan_application['Application_Date'] = pd.to_datetime(loan_application['Application_Date'])
loan_application
```

```python
loan_repayment = pd.read_excel(file, sheet_name = 'Loan_Repayments')
loan_repayment
```

```python
# Identifying NULL Values

loan_repayment.isnull().sum()
```

```python
# Handling NULL Values

loan_repayment['Amount_Paid_INR'] = pd.to_numeric(loan_repayment['Amount_Paid_INR'], errors = 'coerce')
loan_repayment['Amount_Paid_INR'] = loan_repayment['Amount_Paid_INR'].fillna(loan_repayment['Amount_Paid_INR'].median())

loan_repayment['Days_Past_Due'] = pd.to_numeric(loan_repayment['Days_Past_Due'], errors = 'coerce')
loan_repayment['Days_Past_Due'] = loan_repayment['Days_Past_Due'].fillna(loan_repayment['Days_Past_Due'].median())

loan_repayment['Payment_Status'] = loan_repayment['Payment_Status'].fillna(loan_repayment['Payment_Status'].mode()[0])

loan_repayment
```

```python
# Identifying Duplicates 

loan_repayment.duplicated().sum()
```

```python
# Droping Duplicates

loan_repayment.drop_duplicates(inplace = True)
loan_repayment
```

```python
loan_repayment.duplicated().sum()
```

```python
# Handling Inconsistent Data

loan_repayment['Payment_Status'] = loan_repayment['Payment_Status'].replace({
    'Late': 'Late',
    'Missed': 'Missed',
    'OVERDUE': 'Overdue',
    'Paid': 'Paid',
    'Partial': 'Partial'
})

loan_repayment
```

```python
customer_profile = pd.read_excel(file, sheet_name = 'Customer_Profiles')
customer_profile
```

```python
# Identifying NULL values

customer_profile.isnull().sum()
```

```python
# Handling NULL Values

customer_profile['Age'] = pd.to_numeric(customer_profile['Age'], errors = 'coerce')
customer_profile['Age'] = customer_profile['Age'].fillna(customer_profile['Age'].median())

customer_profile['Gender'] = customer_profile['Gender'].fillna(customer_profile['Gender'].mode()[0])

customer_profile['City'] = customer_profile['City'].fillna(customer_profile['City'].mode()[0])

customer_profile['Residence_Type'] = customer_profile['Residence_Type'].fillna(customer_profile['Residence_Type'].mode()[0])

customer_profile['Education_Level'] = customer_profile['Education_Level'].fillna(customer_profile['Education_Level'].mode()[0])

customer_profile['Monthly_Income_INR'] = pd.to_numeric(customer_profile['Monthly_Income_INR'], errors = 'coerce')
customer_profile['Monthly_Income_INR'] = customer_profile['Monthly_Income_INR'].fillna(customer_profile['Monthly_Income_INR'].median())

customer_profile
```

```python
# Identifying Duplicates

customer_profile.duplicated().sum()
```

```python
# Droping Duplicates

customer_profile.drop_duplicates(inplace = True)
customer_profile
```

```python
customer_profile.duplicated().sum()
```

```python
# Handling Inconsistent Data

customer_profile['Gender'] = customer_profile['Gender'].replace({
    'female': 'Female',
    'MALE': 'Male',
    'Others': 'Others'
})

customer_profile['City'] = customer_profile['City'].replace({
    'Aurangabad': 'Aurangabad',
    'mumbai': 'Mumbai',
    'NAGPUR': 'Nagpur',
    'Nashik': 'Nashik',
    'Pune': 'Pune'
})

customer_profile['Residence_Type'] = customer_profile['Residence_Type'].replace({
    'Family': 'Family',
    'Mortgage': 'Mortgage',
    'owned': 'Owned',
    'Rent': 'Rent',
    'Rented': 'Rented'
})

customer_profile['Education_Level'] = customer_profile['Education_Level'].replace({
    'Diploma': 'Diploma',
    'graduate': 'Graduate',
    'High School': 'High School',
    'Other': 'Other',
    'Postgraduate': 'Postgraduate'
})

customer_profile
```

```python
# Saving files in CSV

loan_application.to_csv('Cleaned Loan Application.csv', index = False)
loan_repayment.to_csv('Cleaned Loan Repayment.csv', index = False)
customer_profile.to_csv('Cleaned Customer Profile.csv', index = False)
```
